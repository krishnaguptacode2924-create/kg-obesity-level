```python
import pandas as pd
import joblib

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    classification_report
)
```

```python
df = pd.read_csv("KG_Obesity_Artifact/KG_Obesity_Cleaned.csv")
df.head()
```

```python
X = df.drop('Obesity_Level', axis = 1)
Y = df['Obesity_Level']

X_train, X_test, y_train, y_test = train_test_split( X, Y, test_size=0.2, random_state=42, stratify=Y)
```

```python
rf_model_1 = RandomForestClassifier(n_estimators=200, random_state=42)
rf_model_1.fit(X_train, y_train)
```

```python
y_pred = rf_model_1.predict(X_test)
print("Accuracy:", accuracy_score(y_test, y_pred))
print(classification_report(y_test, y_pred))
print(rf_model_1.feature_names_in_) 
pd.Series(rf_model_1.feature_importances_, index=rf_model_1.feature_names_in_).sort_values(ascending=False)
```

```python
X2 = df.drop(columns = ['Height', 'Weight', 'Obesity_Level'], axis = 1)
Y2 = df['Obesity_Level']

X2_train, X2_test, y2_train, y2_test = train_test_split( X2, Y2, test_size=0.2, random_state=42, stratify=Y2)
```

```python
rf_model_2 = RandomForestClassifier(n_estimators=200, random_state=42)
rf_model_2.fit(X2_train, y2_train)
```

```python
y2_pred = rf_model_2.predict(X2_test)
print("Accuracy:", accuracy_score(y2_test, y2_pred))
print(classification_report(y2_test, y2_pred))
print(rf_model_2.feature_names_in_) 
pd.Series(rf_model_2.feature_importances_, index=rf_model_2.feature_names_in_).sort_values(ascending=False)
```
